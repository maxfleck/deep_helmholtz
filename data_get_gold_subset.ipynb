```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import glob
```

```python
parameter_file = "full_data_6/saft_parameters.csv"

files_path = "full_data_6/data_res_full_*.csv"

files = glob.glob(files_path)
files
```

```python
df = pd.read_csv(parameter_file)
```

```python
df["t_span"] = df["t_max_psat"] - df["t_min_psat"]

gold = (
    (df["mard_psat"] < 3.0)              # fit reproduces psat to better than 3 %
    & (df["points_psat"] >= 20)          # enough psat data to constrain parameters
    & (df["t_span"] >= 100.0)            # over a wide enough T range to separate them
    & (df["bounds_violation"] == "OK")   # optimizer did not hit the parameter box
)
print(f"gold: {gold.sum()} of {len(df)}  ({gold.mean():.1%})")

df = df[gold].reset_index(drop=True)     # <- the gold df
print(df["opt"].value_counts().to_dict())
df.shape
```

```python
ids = np.unique(df.ID)
ids
```

```python
for file in files:
    print(file)
    data = pd.read_csv(file)
    print(data.shape)
    data = data[data.ID.isin(ids)]
    print(data.shape)
    data.to_csv( file.replace(".csv","_gold.csv") , index=False)

data.head()
```
